```python
import os
from collections import defaultdict

extract_path = r'C:\Users\alyss\datasetresize\datasetresized\dataset_resizendenoising'
splits = ["train", "valid", "test"]

for split in splits:
    label_path = os.path.join(extract_path, split, "labels")
    
    class_image_count = defaultdict(set)   # untuk hitung gambar unik per kelas
    class_label_count = defaultdict(int)   # untuk hitung total objek per kelas

    print(f"\n📂 {split.upper()}")

    if os.path.exists(label_path):

        label_files = [f for f in os.listdir(label_path) if f.endswith(".txt")]

        for file in label_files:
            file_path = os.path.join(label_path, file)

            with open(file_path, "r") as f:
                lines = f.readlines()

                classes_in_image = set()

                for line in lines:
                    if line.strip():
                        class_id = line.split()[0]
                        class_label_count[class_id] += 1
                        classes_in_image.add(class_id)

                for cls in classes_in_image:
                    class_image_count[cls].add(file)

        # tampilkan hasil
        for cls in sorted(class_label_count.keys(), key=int):
            print(f"Kelas {cls} : "
                  f"{len(class_image_count[cls])} gambar | "
                  f"{class_label_count[cls]} label")

        print(f"🔢 Total kelas terdeteksi: {len(class_label_count)}")

    else:
        print("Folder labels tidak ditemukan!")
```
